```python
import pandas as pd

df = pd.read_csv("online food delivery dataset-selected-columns.csv")

print("frist 5 rows ")  
print(df.head())

print("\nnumber of rows:", df.shape[0])
print("number of columns:", df.shape[1])


print("\ncolumn names:")
print(df.columns)

print("\ndata types:")
print(df.dtypes)
```

```python
average_age = df["Age"].mean()
print("\nAverage age:", average_age)
minimum_age = df["Age"].min()
print("Minimum age:", minimum_age)
maximum_age = df["Age"].max()
print("Maximum age:", maximum_age)
male_count = df[df["Gender"] == "Male"].shape[0]
female_count = df[df["Gender"] == "Female"].shape[0]
print("\nNumber of males:", male_count)
```

```python
# Problem 3: Filtering Customers

filtered_customers = df[
    (df["Age"] > 25) &
    (df["Occupation"] == "Employee")
]

print(filtered_customers[[
    "Age",
    "Gender",
    "Occupation",
    "Monthly Income"
]])
```

```python
# Problem 4: Occupation Analysis

avg_age_by_occupation = df.groupby("Occupation")["Age"].mean()

print("Average age for each occupation:")
print(avg_age_by_occupation)

highest_avg_age_occupation = avg_age_by_occupation.idxmax()
highest_avg_age = avg_age_by_occupation.max()

print("\nOccupation with highest average age:", highest_avg_age_occupation)
print("Highest average age:", highest_avg_age)
```

```python
# Problem 5: Feedback Analysis

feedback_counts = df["Feedback"].value_counts()

print("Number of customers for each feedback:")
print(feedback_counts)

positive_count = feedback_counts["Positive"]
total_customers = len(df)

positive_percentage = (positive_count / total_customers) * 100

print("\nPercentage of customers who gave positive feedback:",
      positive_percentage)
```

```python
occupation_feedback = df.groupby(["Occupation", "Feedback"]).size()

print(occupation_feedback)
```

```python
class BankAccount:
    
    def __init__(self, account_num, owner, balance):
        self.account_num = account_num
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")
        else:
            print("Insufficient funds.")
```
