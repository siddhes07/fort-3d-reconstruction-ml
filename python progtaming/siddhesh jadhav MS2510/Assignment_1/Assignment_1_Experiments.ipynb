## Experiment 1: Swapping two variables

```
a = 10
b = 20
a, b = b, a
print(a)
print(b)
```

**Explanation:** Python allows swapping two variables in a single line using tuple unpacking: `a, b = b, a`. 
The right-hand side `b, a` is evaluated first (creating a temporary pair), and then assigned to `a, b` on the left. 
So after this line, `a` becomes 20 (old value of b) and `b` becomes 10 (old value of a). No temporary variable is needed, unlike other languages.

## Experiment 2: String Indexing and Slicing

```
text = "Python Programming"
print(text[0])
print(text[-1])
print(text[0:6])
print(text[::-1])
```

**Explanation:**
- `text[0]` gives the first character `'P'` (indexing starts from 0).
- `text[-1]` gives the last character `'g'` (negative index counts from the end).
- `text[0:6]` slices characters from index 0 up to (not including) index 6, giving `'Python'`.
- `text[::-1]` uses a step of -1, which reverses the entire string.

## Experiment 3: List append() and remove()

```
numbers = [10, 20, 30]
numbers.append(40)
numbers.remove(20)
print(numbers)
```

**Explanation:** `append(40)` adds 40 at the end of the list, making it `[10, 20, 30, 40]`. 
Then `remove(20)` deletes the first occurrence of the value 20, giving the final list `[10, 30, 40]`.

## Experiment 4: More String Slicing

```
text = "Programming"
print(text[0:5])
print(text[3:])
print(text[:6])
print(text[::-1])
```

**Explanation:**
- `text[0:5]` gives characters from index 0 to 4 -> `'Progr'`.
- `text[3:]` gives everything from index 3 to the end -> `'gramming'`.
- `text[:6]` gives everything from the start up to index 5 -> `'Progra'`.
- `text[::-1]` reverses the string -> `'gnimmargorP'`.

## Experiment 5: Why input() output differs before and after conversion

```
age = input("Enter your age: ")
print(age + age)
age = int(age)
print(age + age)
# Why are the two outputs different?
```

**Explanation:** `input()` always returns data as a **string**, even if the user types a number. 
So the first `age + age` performs **string concatenation** -- if the user enters `21`, `age + age` gives `'2121'` (the string repeated), not addition.
After `age = int(age)`, the variable is converted to an integer, so the second `age + age` performs actual **mathematical addition** -- `21 + 21 = 42`.

## Experiment 7: Updating a List Element

```
numbers = [10, 20, 30]
numbers[1] = 200
print(numbers)
# Why can a list be modified while a string cannot?
```

**Explanation:** Assigning `numbers[1] = 200` directly changes the element at index 1, giving `[10, 200, 30]`.
This works because **lists are mutable** -- their contents can be changed after creation.
**Strings are immutable** -- once created, their characters cannot be changed in place (e.g. `text[0] = 'X'` would raise an error). 
Instead, any "modification" of a string actually creates a brand new string object.

## Experiment 9: Student Card Program

```
name = input("Enter Name: ")
age = input("Enter Age: ")
roll_no = input("Enter Roll Number: ")
course = input("Enter Course: ")

print("=" * 28)
print("STUDENT CARD")
print("=" * 28)
print("Name    :", name)
print("Age     :", age)
print("Roll No.:", roll_no)
print("Course  :", course)
print("=" * 28)
print()
print(f"Hello {name}!")
print("Welcome to the world of Python.")
print("\"Python is easy to learn!\"")
print("'I\'ll practice every day.'")
print("\"Practice makes perfect.\"")
print("\\/" * 12)
print("Happy Coding!")
```

**Explanation:** This program collects Name, Age, Roll Number, and Course using `input()`, then displays them in a neatly formatted card using `print()` with `=` as a border, mixed with f-strings for the greeting.
Escape characters are used to print special characters literally: `\\"` prints a double quote, and `\\'` prints a single quote inside a string, and `\\\\` prints a backslash.
