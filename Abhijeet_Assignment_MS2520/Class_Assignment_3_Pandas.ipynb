```python
#Question no 9
class BankAccount:

    def __init__(self, account_num, owner, balance):
        self.account_num = account_num
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return True
        else:
            return False


account = BankAccount(101, "Abhijeet", 5000)

result1 = account.withdraw(2000)

print("Valid Withdrawal:", result1)
print("Balance:", account.balance)

result2 = account.withdraw(5000)

print("Invalid Withdrawal:", result2)
print("Balance:", account.balance)
```

```python
#Question no 11

class Student:

    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def average_marks(self):
        return sum(self.marks) / len(self.marks)


student1 = Student("Abhijeet", 1, [80, 85, 90])
student2 = Student("Yash", 2, [75, 80, 85])
student3 = Student("Kunal", 3, [90, 95, 88])


print("Student 1:", student1.name)
print("Average Marks:", student1.average_marks())

print("\nStudent 2:", student2.name)
print("Average Marks:", student2.average_marks())

print("\nStudent 3:", student3.name)
print("Average Marks:", student3.average_marks())
```

```python
#Question no 13
class Vehicle:
    def move(self):
        print("Vehicle is moving")


class Car(Vehicle):
    def move(self):
        print("Car is driving")


class Boat(Vehicle):
    def move(self):
        print("Boat is sailing")


vehicle = Vehicle()
car = Car()
boat = Boat()

vehicle.move()
car.move()
boat.move()
```

```python
#Question no 14
class Vehicle:
    def __init__(self, brand):
        self.brand = brand


class Car(Vehicle):
    def __init__(self, brand, doors):
        super().__init__(brand)
        self.doors = doors


car1 = Car("Mahindra", 4)
car2 = Car("BMW", 4)

print("Car 1")
print("Brand:", car1.brand)
print("Doors:", car1.doors)

print("\nCar 2")
print("Brand:", car2.brand)
print("Doors:", car2.doors)
```
